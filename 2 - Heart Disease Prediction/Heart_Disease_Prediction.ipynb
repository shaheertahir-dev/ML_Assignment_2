| 12 | ca | Number of major blood vessels coloured by fluoroscopy (0-3) |

###  Step 1: Import Libraries

````python
"""
Purpose of this code:
----------------------
This script imports the required libraries for building and evaluating a
machine learning model. It sets up tools for data manipulation, preprocessing,
splitting datasets, training a Support Vector Machine (SVM) model,
evaluating accuracy, and displaying results in tabular form.
"""

# Import numerical computing library (used for arrays, math operations, etc.)
import numpy as np

# Import data manipulation library (used for DataFrames, reading/writing data, etc.)
import pandas as pd

# Import pickle (used for saving and loading Python objects/models)
import pickle

# Import train_test_split (used to divide dataset into training and testing sets)
from sklearn.model_selection import train_test_split

# Import LabelEncoder (used to convert categorical values into numeric codes)
from sklearn.preprocessing import LabelEncoder

# Import StandardScaler (used to put all numbers on the same scale before training)
from sklearn.preprocessing import StandardScaler

# Import make_pipeline (used to join the scaler and the model into one object)
from sklearn.pipeline import make_pipeline

# Import Support Vector Machine (SVM) model from scikit-learn
from sklearn import svm

# Import evaluation metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Import PrettyTable (used to display results in a clean, table format)
from prettytable import PrettyTable

# Import plotting libraries (used for the confusion matrix heatmap)
import matplotlib.pyplot as plt
import seaborn as sns
````

### Step 2: Load Sample Data
The file `processed.cleveland.data` is a comma-separated text file **without a header row** (the first line is already a patient, not column names). So we:
- pass `header=None` to tell pandas there is no header, and
- pass `names=...` to give the 14 columns their names ourselves.

````python
"""
Purpose of this section:
-------------------------
This part of the code loads the heart disease dataset from the data file
into a pandas DataFrame. Because the file has no header row, we add the
14 column names ourselves. Missing values in this file are written as '?',
so we tell pandas to read '?' as a missing value (NaN = "Not a Number").
"""

# The 14 column names of the Cleveland dataset (in the same order as in the file)
column_names = ["age", "sex", "cp", "trestbps", "chol", "fbs", "restecg",
                "thalach", "exang", "oldpeak", "slope", "ca", "thal", "num"]

# Read the data file and load it into a pandas DataFrame
# header=None   -> the file has no header row
# names=...     -> use our 14 column names
# na_values="?" -> treat every '?' in the file as a missing value
sample_data = pd.read_csv("processed.cleveland.data",
                          header=None,
                          names=column_names,
                          na_values="?")

# Print a header to indicate that sample data is being displayed
print("\n\nSample Data (first 10 rows):")
print("============================\n")

# Configure pandas to display all columns (so nothing gets hidden with '...')
pd.set_option("display.max_columns", None)

# Print the first 10 rows of the dataset (303 rows is too long to print in full)
print(sample_data.head(10))
````

### Step 3: Understand and Pre-process Sample Data

####  Step 3.1: Understand Sample Data

````python
"""
Purpose of this section:
-------------------------
This part of the code helps us understand the dataset by:
1. Printing the names of all attributes (columns) in the DataFrame.
2. Showing the total number of instances (rows) available in the dataset.
3. Showing how many values are missing in each column.
4. Showing how many patients have each value of the original target "num".
"""

# Print a header for clarity
print("\n\nAttributes in Sample Data:")
print("==========================\n")

# Display all column names (attributes) in the dataset
print(sample_data.columns)

# Print the number of instances (rows = patients) in the dataset
print("\n\nNumber of Instances in Sample Data:", len(sample_data))
print("========================================\n")

# Count the missing values (the '?' in the file) in every column
print("Missing Values in Each Column:")
print("==============================\n")
print(sample_data.isnull().sum())

# Count how many patients have each value of the original target "num" (0-4)
print("\n\nValues of the Original Target 'num':")
print("====================================\n")
print(sample_data["num"].value_counts().sort_index())
````

#### Step 3.2: Pre-process Sample Data
o   **Missing values:** The `?` values were already turned into missing values (NaN) when we loaded the file. Step 3.1 shows they are **only in the `ca` and `thal` columns**. We do **not** use these two columns, so we do not need to fill them in. To be safe, Step 4 still removes any row that has a missing value in the columns we do use.

o   **Binary target:** `num` has 5 values (0, 1, 2, 3, 4). We only want to know *disease or no disease*, so we make a new column `target`: **0 = no disease** (num is 0) and **1 = disease** (num is 1, 2, 3 or 4).

o   **Readable names:** `sex` and `cp` are stored as code numbers (for example cp = 4). We change them to readable words (for example "Asymptomatic"), just like the Titanic data used words. This makes the data easier to read, and lets a user type words in Step 8.

````python
"""
Purpose of this section:
-------------------------
This part of the code pre-processes the data:
1. Turns the target "num" (0-4) into a binary target (0 = no disease, 1 = disease).
2. Changes the code numbers in "sex" and "cp" into readable words.
"""

# Make the binary target: True if num is greater than 0 (disease), False if num is 0
# .astype(int) turns True/False into 1/0
sample_data["target"] = (sample_data["num"] > 0).astype(int)

# Show that the conversion is correct: each num value and the target it became
print("\n\nOriginal 'num' vs New Binary 'target':")
print("======================================\n")
print(sample_data.groupby("num")["target"].first())

# Count patients in each class of the new target
print("\n\nNumber of Patients per Class (0 = No Disease, 1 = Disease):")
print("===========================================================\n")
print(sample_data["target"].value_counts().sort_index())

# Dictionary that says which word belongs to which sex code
sex_names = {1: "Male", 0: "Female"}

# Dictionary that says which word belongs to which chest pain (cp) code
cp_names = {1: "Typical Angina", 2: "Atypical Angina",
            3: "Non-anginal Pain", 4: "Asymptomatic"}

# Replace the code numbers with the words, using the dictionaries above
sample_data["sex"] = sample_data["sex"].map(sex_names)
sample_data["cp"] = sample_data["cp"].map(cp_names)

# Show the first 5 rows after pre-processing
print("\n\nSample Data after Pre-processing (first 5 rows):")
print("================================================\n")
print(sample_data.head())
````

#### Step 4: Feature Extraction
We keep only **4 input attributes** and the **target**. Here is why each input is useful for predicting heart disease:

| Attribute | What it is | Why it is useful |
|---|---|---|
| **age** | Age in years | The risk of heart disease goes up as people get older, because arteries become stiffer and narrower over time. In this data, patients with disease are on average about 4 years older (56.6 vs 52.6). |
| **sex** | Male or Female | Men get heart disease more often, and at a younger age, than women. In this data 55% of men have disease but only 26% of women. |
| **cp** | Chest pain type | Chest pain is the main symptom doctors ask about. Surprisingly, patients with **Asymptomatic** chest pain (no typical pain) are the most likely to have disease (73%), compared with 18-30% for the other three types, so the *type* of pain tells the model a lot. |
| **thalach** | Maximum heart rate during an exercise test | A healthy heart can beat faster during exercise. A **low** maximum heart rate can mean the heart is not getting enough blood, which is a sign of disease. In this data, patients with disease reach about 139 beats per minute on average, healthy patients about 158. |

All four are also **easy to collect**: age and sex are known, chest pain is described by the patient, and the maximum heart rate comes from a simple exercise test. The other 9 columns need more tests (blood tests, ECG, scans), and two of them (`ca`, `thal`) also have missing values.

````python
"""
Purpose of this section:
-------------------------
This part of the code keeps only the 4 input attributes and the output
(target), and removes any row that still has a missing value in them.
"""

# The 4 input attributes we chose, plus the output column "target" at the end
selected_columns = ["age", "sex", "cp", "thalach", "target"]

# Keep only the selected columns (.copy() makes a new, separate DataFrame)
sample_data = sample_data[selected_columns].copy()

# Remember how many rows we have before removing missing values
rows_before = len(sample_data)

# Remove any row that has a missing value in the selected columns
sample_data = sample_data.dropna()

# Print how many rows were removed (0 means no missing values in our columns)
print("Rows removed because of missing values:", rows_before - len(sample_data))
print("Number of Instances after Feature Extraction:", len(sample_data))

# Print the first 10 rows of the data we will use
print("\n\nSample Data after Feature Extraction (first 10 rows):")
print("=====================================================\n")
print(sample_data.head(10))
````

### Step 5: Label Encoding the Sample Data (Input and Output is converted in Numeric Representation)
Machine learning models only work with numbers.
- **sex** and **cp** are words now, so we turn them into numbers with a `LabelEncoder` (just like the Titanic notebook).
- **age** and **thalach** are already numbers, so they do not need label encoding.
- **target** is already a number (0 or 1) from Step 3.2.

#### Step 5.1: Train the Label Encoder

````python
"""
Purpose of this section:
-------------------------
This part of the code sets up and trains Label Encoders.
Label Encoding is used to convert categorical values (like 'Male', 'Female',
'Asymptomatic') into numeric codes that machine learning algorithms can work with.

Steps:
1. Define the possible categories for each word attribute.
2. Initialize a LabelEncoder for each attribute.
3. Train (fit) each LabelEncoder with its respective category values.
"""

# Define label categories for each word attribute
# These are all the possible values each column can have in the dataset
sex = pd.DataFrame({"Sex": ["Male", "Female"]})
cp = pd.DataFrame({"Cp": ["Typical Angina", "Atypical Angina", "Non-anginal Pain", "Asymptomatic"]})

# Create LabelEncoder objects for each word column
sex_label_encoder = LabelEncoder()
cp_label_encoder = LabelEncoder()

# Train (fit) the encoders on the given categories
# np.ravel() flattens the DataFrame column into a 1D array for the encoder
sex_label_encoder.fit(np.ravel(sex))
cp_label_encoder.fit(np.ravel(cp))

# Show which number each word gets (LabelEncoder sorts the words alphabetically)
# enumerate() gives each word its position number: 0, 1, 2, ...
print("Sex codes:")
for code, name in enumerate(sex_label_encoder.classes_):
    print("  ", code, "=", name)
print("Cp codes:")
for code, name in enumerate(cp_label_encoder.classes_):
    print("  ", code, "=", name)
````

#### Step 5.2: Label Encode the Output

````python
"""
Purpose of this section:
-------------------------
This part of the code deals with the target/output variable ("target").
It was already converted into numbers in Step 3.2 (0 = no disease,
1 = disease), so no Label Encoder is needed. We just check the values
and save this version of the dataset.
"""

# Create a copy of the dataset (so the original stays unchanged)
sample_data_encoded_output = sample_data.copy()

# Show the possible output values (they should only be 0 and 1)
print("\n\nTarget Attribute (already numeric):")
print("===================================\n")
print("Unique values in target:", sorted(sample_data_encoded_output["target"].unique().tolist()))

# Show the first 10 output values
print(sample_data_encoded_output[["target"]].head(10))

# Save this dataset to a CSV file (useful for checking later)
sample_data_encoded_output.to_csv(r'sample-data-encoded-output.csv', index=False, header=True)
print("\nSaved: sample-data-encoded-output.csv")
````

#### Step 5.3: Label Encode the Input

````python
"""
Purpose of this section:
-------------------------
This part of the code applies Label Encoding to the word input features
("sex", "cp"). Each word is converted into a number so that machine
learning models can process it. "age" and "thalach" stay as they are.
"""

# Create a copy of the dataset (to keep the word version safe)
sample_data_encoded = sample_data_encoded_output.copy()

# Encode each word attribute into numbers using the trained encoders
sample_data_encoded["sex"] = sex_label_encoder.transform(sample_data_encoded["sex"])
sample_data_encoded["cp"] = cp_label_encoder.transform(sample_data_encoded["cp"])

# Print encoded dataset (with numbers instead of words)
print("\n\nSample Data after Label Encoding (first 10 rows):")
print("=================================================\n")
print(sample_data_encoded.head(10))

# Save the encoded dataset to a new CSV file
sample_data_encoded.to_csv(r'sample-data-encoded.csv', index=False, header=True)
print("\nSaved: sample-data-encoded.csv")
````

### Step 6: Execute the Training Phase

#### Step 6.1: Splitting Sample Data into Training Data and Testing Data
We keep **80%** of the patients for training and hide **20%** for testing. We use `shuffle=True` so the rows are mixed before splitting (we don't know how the rows in this file were ordered), and `random_state=0` so we get the same split every time we run the notebook.

````python
"""
Purpose of this section:
-------------------------
This part of the code splits the dataset into Training Data and Testing Data.
- Training Data is used to teach the machine learning model.
- Testing Data is used to check how well the model performs on unseen data.
"""

# Split dataset into training (80%) and testing (20%)
# shuffle=True mixes the rows before splitting
# random_state=0 ensures reproducibility (same split every time)
training_data_encoded, testing_data_encoded = train_test_split(
    sample_data_encoded,
    test_size=0.2,
    random_state=0,
    shuffle=True
)

# Save the training and testing datasets to CSV files
training_data_encoded.to_csv(r'training-data-encoded.csv', index=False, header=True)
testing_data_encoded.to_csv(r'testing-data-encoded.csv', index=False, header=True)

# Print how many rows and columns each part has
print("Training Data shape:", training_data_encoded.shape)
print("Testing Data shape:", testing_data_encoded.shape)
````

### Step 6.2: Splitting Input Vectors and Outputs / Labels of Training Data

````python
"""
Purpose of this section:
-------------------------
This part of the code separates the features (input vectors) and
the labels (output values) from the training dataset.
"""

# Extract input vectors (features) from training data: all columns except the last one
input_vector_train = training_data_encoded.iloc[:, :-1]

# Extract output labels (target variable) from training data: only the last column
output_label_train = training_data_encoded.iloc[:, -1]

# Print the size of the inputs and outputs
print("Input vectors (train) shape:", input_vector_train.shape)
print("Output labels (train) shape:", output_label_train.shape)
````

### 6.3: Train the Support Vector Classifier
**Why we add a `StandardScaler`:** In the Titanic notebook every input was a small code (0, 1, 2, 3). Here, `age` (29-77) and `thalach` (71-202) are much bigger numbers than `sex` (0-1) and `cp` (0-3). A Support Vector Classifier measures *distances* between patients, so the big numbers would drown out the small ones. A `StandardScaler` puts every input on the same scale (average 0, spread 1) so all 4 inputs count fairly.

We join the scaler and the SVC into **one pipeline**. This way the saved model always scales new data the same way before predicting, and we only need to save one `.pkl` file.

````python
"""
Purpose of this section:
-------------------------
This part of the code trains a Support Vector Classifier (SVC) on the training data.
The SVC is a type of machine learning model that finds the best boundary
to separate different classes (e.g., disease vs. no disease).
"""

print("\n\nTraining the Support Vector Classifier on Training Data")
print("========================================================\n")

# Create the model: first scale the inputs, then use a Support Vector Classifier
# random_state=0 ensures the same result every time
svc_model = make_pipeline(StandardScaler(), svm.SVC(random_state=0))

# Train the model using the training data (features and labels)
svc_model.fit(input_vector_train, np.ravel(output_label_train))

# Print trained model details (the steps in the pipeline)
print(svc_model)
````

### Step 6.4: Save the Trained Model

````python
"""
Purpose of this section:
-------------------------
This part of the code saves the trained model (scaler + SVC) to your
computer so it can be reused later without retraining.
"""

# Save the trained model to a file named "heart_disease_svc_model.pkl"
# 'wb' means "write in binary mode"; the with-block closes the file for us
with open('heart_disease_svc_model.pkl', 'wb') as model_file:
    pickle.dump(svc_model, model_file)
print("Saved: heart_disease_svc_model.pkl")
````

### Step 7: Execute the Testing Phase

#### Step 7.1: Splitting Input Vectors and Outputs/Labels of Testing Data

````python
"""
Purpose of this section:
-------------------------
This part of the code separates the features (inputs) and the labels (outputs)
from the testing dataset.
"""

# Inputs of the testing data: all columns except the last one
input_vector_test = testing_data_encoded.iloc[:, :-1]

# Outputs (true answers) of the testing data: only the last column
output_label_test = testing_data_encoded.iloc[:, -1]

# Print the size of the inputs and outputs
print("Input vectors (test) shape:", input_vector_test.shape)
print("Output labels (test) shape:", output_label_test.shape)
````

### Step 7.2: Load the Saved Model

````python
"""
Purpose of this section:
-------------------------
This part of the code loads the previously saved model from disk into memory.
"""

# Open the saved model file in 'rb' (read binary) mode and load the model
with open('heart_disease_svc_model.pkl', 'rb') as model_file:
    model = pickle.load(model_file)
print("Model loaded from heart_disease_svc_model.pkl")
````

### Step 7.3: Evaluate the Machine Learning Model

#### Step 7.3.1: Make Predictions with the Trained Models on Testing Data

````python
"""
Purpose of this section:
-------------------------
This part of the code uses the trained model to make predictions
on the testing data (patients the model has never seen).
"""

# Ask the model to predict disease (1) or no disease (0) for each test patient
model_predictions = model.predict(input_vector_test)

# Make a copy of the testing data so we can add a "Predictions" column to it
model_predictions_table = testing_data_encoded.copy()

# Add the model's predictions next to the true answers
model_predictions_table["Predictions"] = model_predictions

# Save the true answers and predictions to a CSV file
model_predictions_table.to_csv(r'model-predictions.csv', index=False, header=True)

# Print the true answers (target) and the model's predictions side by side
print("\n\nPredictions Returned by heart_disease_svc_model:")
print("================================================\n")
print(model_predictions_table)
````

### Step 7.4: Calculate the Accuracy Score

````python
"""
Purpose of this section:
-------------------------
This part of the code calculates how accurate the trained model is on the
testing dataset (the share of test patients it predicted correctly).
"""

# Compare the true answers with the predictions and get the accuracy
model_accuracy_score = accuracy_score(
    model_predictions_table["target"],
    model_predictions_table["Predictions"]
)

print("\n\nAccuracy Score:")
print("===============\n")

# Print the accuracy rounded to 2 decimal places (1.0 would mean 100% correct)
print(round(model_accuracy_score, 2))
````

### Step 7.5: Detailed Evaluation Metrics (Precision, Recall, F1, Confusion Matrix)

````python
"""
Purpose of this section:
-------------------------
Accuracy alone can be misleading. This section adds the standard
classification metrics:
- Precision: of everyone the model predicted as "Disease", how many really had disease?
- Recall: of everyone who really had disease, how many did the model catch?
- F1-score: the harmonic mean of precision and recall.
- Confusion Matrix: a table of correct vs. incorrect predictions per class.
For a medical test, Recall for "Disease" is very important: a missed sick
patient is usually worse than a healthy patient sent for an extra test.
"""

print("\nClassification Report:")
print("=======================\n")

# Print precision, recall and F1-score for each class
print(classification_report(
    model_predictions_table["target"],
    model_predictions_table["Predictions"],
    target_names=["No Disease (0)", "Disease (1)"]
))

# Build the confusion matrix (rows = true answers, columns = predictions)
conf_matrix = confusion_matrix(model_predictions_table["target"], model_predictions_table["Predictions"])

# Put the confusion matrix in a DataFrame with clear row and column names
conf_matrix_df = pd.DataFrame(
    conf_matrix,
    index=["Actual: No Disease", "Actual: Disease"],
    columns=["Predicted: No Disease", "Predicted: Disease"]
)
print("\nConfusion Matrix:")
print("==================\n")
print(conf_matrix_df)

# Draw the confusion matrix as a coloured heatmap (darker = more patients)
plt.figure(figsize=(5, 4))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=['No Disease', 'Disease'],
            yticklabels=['No Disease', 'Disease'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix - SVC on Heart Disease Testing Data')
plt.tight_layout()
plt.show()
````
